```python
import pandas as pd
import numpy as np 

import joblib
```

```python
synthetic_data = pd.DataFrame({
    "pH": [7.55, 7.55, 7.55, 7.55, 7.55],
    "TDS": [750, 920, 800, 800, 800],
    "Turbidity": [2.0, 2.0, 2.0, 2.0, 2.0],
    "Phospate": [0.010, 0.010, 0.003, 0.010, 0.010],
    "Nitrate": [7.0, 7.0, 7.0, 7.0, 7.0],
    "Iron": [0.35, 0.35, 0.35, 0.42, 0.35],
    "COD(mg/L)": [405, 395, 395, 395, 395],
    "Chlorine": [2.0, 2.0, 2.0, 2.0, 2.0],
    "Sodium": [10.0, 10.0, 10.0, 10.0, 10.0]
})

synthetic_data
```

```python
def discovered_rule(row):
    if row["COD(mg/L)"] > 399.5:
        return 3
    elif row["TDS"] > 899.5:
        return 2
    elif row["Phospate"] <= 0.005:
        return 1
    elif row["Iron"] > 0.4:
        return 0
    else:
        return 4
```

```python
logistic_pipeline = joblib.load("../models/logistic_regression.joblib")
decision_tree_pipeline = joblib.load("../models/decision_tree.joblib")
rf_pipeline = joblib.load("../models/random_forest.joblib")
xgb_pipeline = joblib.load("../models/xgboost.joblib")
cat_pipeline = joblib.load("../models/catboost.joblib")
```

```python
synthetic_predictions = pd.DataFrame({
    "Logistic Regression": logistic_pipeline.predict(synthetic_data),
    "Decision Tree": decision_tree_pipeline.predict(synthetic_data),
    "Random Forest": rf_pipeline.predict(synthetic_data),
    "XGBoost": xgb_pipeline.predict(synthetic_data),
    "CatBoost": cat_pipeline.predict(synthetic_data).ravel()
})
```

```python
synthetic_data["EDA Rule"] = synthetic_data.apply(discovered_rule, axis=1)
```

```python
synthetic_results = pd.concat(
    [synthetic_data, synthetic_predictions],
    axis=1
)

synthetic_results
```
